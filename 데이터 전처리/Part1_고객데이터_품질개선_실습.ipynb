- 결측치, 중복값, 이상치와 범주 표기 불일치를 분석 목적에 맞게 처리합니다.

### 4-1. 데이터 품질 문제의 영향과 한계

- 선택한 품질 문제: 결측치와 이상치가 존재하여 결측치가 있는 행을 삭제하여 처리했다. Age, VisitCount, AveragePurchaseAmount 등이 처음에는 문자열로 저장되어 있어 수치형으로 변환하였다.

- EDA 또는 모델링에 미치는 영향: 결측치가 있는 데이터를 사용하면 결과값이 왜곡될 수 있어 모델링 과정에서도 예측 성능에 영향을 줄 수 있다.

- 추가 확인이 필요한 조건 또는 한계: 결측치가 발생한 원인이 명확하지 않다. 또한 이상치로 판단된 값도 실제 오류인지, 구매 금액이 높은 정상적인 고객인지 판단하기 어렵다. 따라서 결측치 발생 원인, 이상치의 실제 거래 여부, 데이터 수집 기간 등을 확인해야한다.

### 4-2. 결측치·이상치 처리 방법의 선택 근거

- 결측치 처리 방법: 결측치가 포함된 행을 삭제했다.(dropna())
- 해당 방법을 선택한 이유: 결측치가 여러 컬럼에 존재했으며 결측값을 처리 안하면 결과 데이터가 왜곡될 수 있기 때문이다.
- 이상치 처리 방법: 수치형 주요 컬럼에 대해 IQR기준을 적용해 이상치를 탐지했고 , 하한과 상한을 벗어나는 값은 제거했다.
- 이상치 처리 시 주의점: 이상치가 반드시 오류 데이터는 아닐 수 있다. 실제로 구매 금액이 높은 고객이거나 방문 횟수가 높은 고객일 수 있기 때문에 확인후 이상치로 처리해야한다.

### 4-3. 파생변수의 활용 의도와 한계

- `AgeGroup` 생성 기준과 활용 목적: AgeGroup은 정제한 Age를 0~29, 29 초과~39, 39 초과~49, 49 초과~59, 59 초과 구간으로 나누고 20대 이하·30대·40대·50대·60대 이상으로 이름 붙였다.
- `PurchaseGrade` 생성 기준과 활용 목적: PurchaseGrade는 정제한 총구매금액의 Q1 186350 이하를 Low, Q1 초과 Q3 542200 이하를 Medium, Q3 초과를 High로 정했다
- 파생변수 사용 시 한계: 구간의 기준에 따라 고객 분류 결과가 달라질 수 있으며, 같은 구간에 속한 고객이라도 실제 구매금액이나 연령 차이가 클 수 있다.
